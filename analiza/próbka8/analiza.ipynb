```python
#waterfall plot dla pojedynczych widm ramanowskich 
x_data = [rs_Raman_0s[0],rs_Raman_2s[12],rs_Raman_4s[12],rs_Raman_6s[12],rs_Raman_8s[12],rs_Raman_10s[12]]
y_data = [I_Raman_0s[7],I_Raman_2s[12],I_Raman_4s[12],I_Raman_6s[12],I_Raman_8s[12],I_Raman_10s[12]]
labels = ["t=0s","t=2s","t=4s","t=6s","t=8s","t=10s"]

y_offset_step=2000
x_offset_step=0

plt.figure(figsize=(8, 6))

# Iteracja po danych i aplikowanie przesunięcia
for i, (x,y,label,color) in enumerate(zip(x_data,y_data,labels,colors)):
    x_shifted=x+(i*x_offset_step)
    y_shifted=y +(i*y_offset_step)
    
    plt.plot(x_shifted,y_shifted,label=label,color=color)

plt.yticks([], []) #ukrycie jednostek na osi y


plt.xlabel(r"Ramanshift (cm$^{-1}$)")
plt.ylabel("Intensity (a.u.)")
plt.axvline(248.22, color=colors[0],linestyle="--") #średnia pozycja piku E+A
plt.axvline(260.34, color=colors[0],linestyle="--")#średnia pozycja piku 2LA(M)

plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left', borderaxespad=0.)
plt.xlim(230,270)
plt.savefig(sciezka/'Raman_pojedyncze_widma.png')
```
