```python
x = np.linspace(1.45, 1.85, 500)
time = [0, 30, 60, 90, 120, 150, 180, 210, 240, 270, 300]

for t, c, f, h, col in zip(time, centers_PL_mean_9, fwhms_PL_mean_9, intensity_PL_mean_9, colors):
    #jeśli się nie dopasowało to fit zwraca nan, pozbywamy się ich 
    mask = ~np.isnan(c) & ~np.isnan(f) & ~np.isnan(h)
    c0 = np.mean(np.array(c)[mask])
    f0 = np.mean(np.array(f)[mask])
    h0 = np.mean(np.array(h)[mask])
    y = lorentz(x, h0, c0, f0)
    plt.plot(x, y, label=f"t={t}s", color=col)

plt.xlabel("E (eV)")
plt.ylabel("Intensity (a.u.)")
plt.xlim(1.45, 1.85)
plt.legend()
plt.savefig(sciezka/'usrednione_PL.png')
```
